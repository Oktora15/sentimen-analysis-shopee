```python
import pandas as pd
from google_play_scraper import reviews, Sort

result, _ = reviews(
    'com.shopee.id',
    lang='id',
    country='id',
    sort=Sort.NEWEST,
    count=3000
)

data = pd.DataFrame(result)
data = data[['content', 'score']]

data.head()
```

```python
data.to_csv('dataset.csv', index=False)
```
